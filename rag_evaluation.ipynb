# RAG Evaluation Framework – Prototype 1

This notebook implements a minimal, controlled evaluation framework for
retrieval-augmented generation (RAG) systems.

We begin with a small synthetic corpus to:
- Create a controlled retrieval task
- Establish ground-truth mappings
- Enable systematic evaluation of retrieval accuracy and answer faithfulness

## Environment Setup

This cell loads environment variables and verifies that required configuration

```python
from dotenv import load_dotenv
load_dotenv()

print("Environment variables loaded.")
```

## Step 1 — Define Controlled Corpus and Queries

We define:
- A small document corpus
- A set of queries
- Ground-truth document IDs for retrieval evaluation

```python
documents = [
    {
        "id": 1,
        "text": "Anthropic develops AI systems with a focus on safety and alignment research."
    },
    {
        "id": 2,
        "text": "Retrieval-Augmented Generation (RAG) improves factual accuracy by incorporating external documents."
    },
    {
        "id": 3,
        "text": "Hallucinations in large language models occur when the model generates unsupported or fabricated information."
    },
    {
        "id": 4,
        "text": "Vector databases enable similarity search over embedded representations of text."
    }
]
```

```python
queries = [
    {
        "query": "What causes hallucinations in language models?",
        "ground_truth_doc_id": 3
    },
    {
        "query": "How does RAG improve factual accuracy?",
        "ground_truth_doc_id": 2
    },
    {
        "query": "What is the role of vector databases in RAG?",
        "ground_truth_doc_id": 4
    }
]
```

## Step 2 — Generate Document Embeddings

We will use the `EmbeddingModel` from `src/embeddings.py` to generate vector representations
for each document in our synthetic corpus. This allows us to perform similarity search
and RAG-style retrieval later.


```python
# Import our embedding abstraction
from rag_eval.embeddings import EmbeddingModel

# Initialize the embedding model
embed_model = EmbeddingModel(model_name="text-embedding-3-small")

# Extract document texts
texts = [doc["text"] for doc in documents]

# Generate embeddings
embeddings = embed_model.embed_texts(texts)

# Attach embeddings to the documents
for doc, emb in zip(documents, embeddings):
    doc["embedding"] = emb

# Verify
documents[0]  # shows first doc with its embedding
```
